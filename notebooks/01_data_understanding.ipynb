Import Libraries

```python
import pandas as pd
import numpy as np
import os
import warnings

warnings.filterwarnings("ignore")

print("Libraries imported successfully!")
```

Define Dataset Path

```python
DATA_PATH = "../data/raw/"

print("Dataset path:", DATA_PATH)
print("Path exists:", os.path.exists(DATA_PATH))
```

Check Available Files

```python
files = os.listdir(DATA_PATH)

print("Files found:")
for file in files:
    print("-", file)
```

Define File Paths
(Instead of repeatedly writing long paths, define them once.)

```python
CUSTOMERS_FILE = DATA_PATH + "olist_customers_dataset.csv"
GEOLOCATION_FILE = DATA_PATH + "olist_geolocation_dataset.csv"
ORDER_ITEMS_FILE = DATA_PATH + "olist_order_items_dataset.csv"
PAYMENTS_FILE = DATA_PATH + "olist_order_payments_dataset.csv"
REVIEWS_FILE = DATA_PATH + "olist_order_reviews_dataset.csv"
ORDERS_FILE = DATA_PATH + "olist_orders_dataset.csv"
PRODUCTS_FILE = DATA_PATH + "olist_products_dataset.csv"
SELLERS_FILE = DATA_PATH + "olist_sellers_dataset.csv"
TRANSLATION_FILE = DATA_PATH + "product_category_name_translation.csv"
```

Load All Datasets

```python
customers = pd.read_csv(CUSTOMERS_FILE)
geolocation = pd.read_csv(GEOLOCATION_FILE)
order_items = pd.read_csv(ORDER_ITEMS_FILE)
payments = pd.read_csv(PAYMENTS_FILE)
reviews = pd.read_csv(REVIEWS_FILE)
orders = pd.read_csv(ORDERS_FILE)
products = pd.read_csv(PRODUCTS_FILE)
sellers = pd.read_csv(SELLERS_FILE)
translation = pd.read_csv(TRANSLATION_FILE)

print("All datasets loaded successfully!")
```

Create Dataset Dictionary

```python
datasets = {
    "customers": customers,
    "geolocation": geolocation,
    "order_items": order_items,
    "payments": payments,
    "reviews": reviews,
    "orders": orders,
    "products": products,
    "sellers": sellers,
    "translation": translation
}

print(f"Total datasets loaded: {len(datasets)}")
```

Missing Values

```python
missing_summary = {}

for name, df in datasets.items():
    missing = df.isnull().sum()
    missing = missing[missing > 0]
    
    missing_summary[name] = missing

    print("\n" + "=" * 70)
    print(name.upper())
    print("=" * 70)
    
    if len(missing) == 0:
        print("No missing values.")
    else:
        print(missing)
```

Missing Value Percentage
 (Counts are useful, but percentages are better for data-quality analysis)

```python
for name, df in datasets.items():
    
    missing_count = df.isnull().sum()
    missing_percentage = (missing_count / len(df)) * 100
    
    missing_table = pd.DataFrame({
        "Missing Count": missing_count,
        "Missing %": missing_percentage.round(2)
    })
    
    missing_table = missing_table[
        missing_table["Missing Count"] > 0
    ]
    
    print("\n" + "=" * 70)
    print(name.upper())
    print("=" * 70)
    
    if missing_table.empty:
        print("No missing values.")
    else:
        display(missing_table)
```

Convert Date Columns
(start preparing the data for analytics)

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders[col] = pd.to_datetime(
        orders[col],
        errors="coerce"
    )

print(orders[date_columns].dtypes)
```

Create Basic Delivery Metrics
(Now calculate delivery time.)

```python
orders["delivery_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_purchase_timestamp"]
).dt.total_seconds() / (60 * 60 * 24)

orders["delivery_days"].describe()
```

Calculate Late Delivery

```python
#Compare actual delivery with estimated delivery.

orders["is_late"] = (
    orders["order_delivered_customer_date"]
    > orders["order_estimated_delivery_date"]
)

orders["is_late"].value_counts()
```

```python
#Percentage

late_percentage = orders["is_late"].mean() * 100

print(f"Late delivery percentage: {late_percentage:.2f}%")
```

Basic Revenue Calculation

```python
#Revenue comes from order_items

total_revenue = order_items["price"].sum()

print(f"Total Revenue: {total_revenue:,.2f}")
```

```python
#Freight

total_freight = order_items["freight_value"].sum()

print(f"Total Freight: {total_freight:,.2f}")
```

Number of Orders and Customers

```python
total_orders = orders["order_id"].nunique()
total_customers = customers["customer_unique_id"].nunique()

print(f"Total Orders: {total_orders:,}")
print(f"Total Customers: {total_customers:,}")
```

Average Order Value

```python
aov = total_revenue / total_orders

print(f"Average Order Value: {aov:,.2f}")
```

Create Initial KPI Summary

(Now combine everything we've learned.)

```python
kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Revenue",
        "Total Orders",
        "Total Customers",
        "Average Order Value",
        "Total Freight",
        "Average Delivery Days",
        "Late Delivery %"
    ],
    
    "Value": [
        total_revenue,
        total_orders,
        total_customers,
        aov,
        total_freight,
        orders["delivery_days"].mean(),
        late_percentage
    ]
})

kpi_summary
```

Create Data Quality Summary


This will eventually become part of our Data Quality Agent.

```python
quality_summary = []

for name, df in datasets.items():
    
    total_cells = df.shape[0] * df.shape[1]
    missing_cells = df.isnull().sum().sum()
    duplicate_rows = df.duplicated().sum()
    
    quality_summary.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing Cells": missing_cells,
        "Missing %": round(
            (missing_cells / total_cells) * 100, 2
        ),
        "Duplicate Rows": duplicate_rows
    })

quality_summary = pd.DataFrame(quality_summary)

quality_summary
```

Save Initial Summary

```python
import os

processed_path = "../data/processed"

os.makedirs(processed_path, exist_ok=True)

print("Processed folder ready:", os.path.exists(processed_path))
```

```python
quality_summary.to_csv(
    "../data/processed/data_quality_summary.csv",
    index=False
)

kpi_summary.to_csv(
    "../data/processed/kpi_summary.csv",
    index=False
)

print("Summary files saved successfully!")
```
